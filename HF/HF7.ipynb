```python
# Open JSON file

file_path = r'C:\Users\Ádám\Desktop\DATA_ANALYST_FILES\CUBIX_Data_Engineer\week7\HF\spotify_playlist.json'
with open(file_path,"r") as file:
    data = json.load(file)
```

```python
#Print out the 31st track's name
req_song_name = data['contents']['items'][30]['name']

print(req_song_name)
```

```python
#Print out total playcont for 50 tracks

total_count = 0

for track in data['contents']['items']:
    total_count = total_count + track['playCount']
print(total_count)
```

```python
# Track with the lowest playcount

min_playcount = float('inf')

for track in data['contents']['items']:
    if track['playCount'] < min_playcount:
        min_playcount = track['playCount']
        track_name = track['name']

print(f'The song name: {track_name}')
```
