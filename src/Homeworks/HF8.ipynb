```python
    """
    Which 10 companies make the most money?
    Show two columns:
    •Company name
    •Sum of the fare (make sure the fare total is not in scientific notation)
    
    Sort the results by total fare, in descending order
    """


SELECT dc.company , ROUND(SUM(ftt.fare),2) as total_fare
FROM chicago_taxi_db.fact_taxi_trips ftt
LEFT JOIN dim_company dc ON dc.company_id = ftt.company_id
GROUP BY dc.company
ORDER BY total_fare desc
LIMIT 10 
```

```python
    """
    Top 10 pickup community areas with the most rides
    Show two columns:
    •Community area name
    •Count of rides
    
    Sort by ride count in descending order.
    """



SELECT  dca.community_area AS Community_area_name , COUNT(ftt.trip_id) AS Count_of_rides
FROM chicago_taxi_db.fact_taxi_trips ftt 
LEFT JOIN dim_community_areas dca ON dca.area_code = ftt.pickup_community_area_id
GROUP BY dca.community_area
ORDER BY Count_of_rides DESC
LIMIT 10
```

```python
    """
    Count taxi rides per day of the week
    •Show the day of the week by name (Monday, Tuesday, etc.)
    •Count how many rides happened on each day
    •Use CASE WHENto convert day numbers (1-7) to names
    •Order results by actual weekday order (starting from Monday)
    """


SELECT 
CASE
    WHEN DAY_OF_WEEK(trip_end_timestamp) = 1 THEN 'Monday'
    WHEN DAY_OF_WEEK(trip_end_timestamp) = 2 THEN 'Tuesday'
    WHEN DAY_OF_WEEK(trip_end_timestamp) = 3 THEN 'Wednesday'
    WHEN DAY_OF_WEEK(trip_end_timestamp) = 4 THEN 'Thursday'
    WHEN DAY_OF_WEEK(trip_end_timestamp) = 5 THEN 'Friday'
    WHEN DAY_OF_WEEK(trip_end_timestamp) = 6 THEN 'Saturday'
    WHEN DAY_OF_WEEK(trip_end_timestamp) = 7 THEN 'Sunday'
END AS Day_of_week, 
COUNT(trip_id) AS No_of_trips
FROM chicago_taxi_db.fact_taxi_trips
GROUP BY DAY_OF_WEEK(trip_end_timestamp)
ORDER BY DAY_OF_WEEK(trip_end_timestamp)
```
