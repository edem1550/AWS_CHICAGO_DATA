```python
import requests
import pandas as pd
# Azért, hogy pd.Dataframe-nél ne csak az általa választott oszlopmennyiséget adja meg, ezért szükség van az alábbira:
pd.set_option("display.max_columns",30)
from datetime import datetime

# napok közit különbség kiszámolása
from dateutil.relativedelta import relativedelta
```

```python
current_datetime = datetime.now() - relativedelta(months=2)

formatted_datetime = current_datetime.strftime("%Y-%m-%d")

formatted_datetime
```

```python
#taxi trips lekérés fix napra:

url = (
    f"https://data.cityofchicago.org/resource/ajtu-isnz.json?"
    f"$where=trip_start_timestamp >= '{formatted_datetime}T00:00:00'"
    f"AND trip_start_timestamp <= '{formatted_datetime}T23:59:59'"
    f"&$limit=30000"
)

response = requests.get(url)
data = response.json()

len(data)
```

```python
# Adatok df-esítése

taxi_trips = pd.DataFrame(data)

taxi_trips.head()
```

```python
# Adatok általános vizsgálata

taxi_trips.describe()
```

## Adatok átalakítása

## NaN adatok elhagyása

```python
taxi_trips.drop(['pickup_census_tract' , 'dropoff_census_tract'], axis=1, inplace=True)
```

```python
# A pickup centroid latitude + longitude = pickup_centroid_location --> duplikátum --> nem kell

taxi_trips.drop(['pickup_centroid_location','dropoff_centroid_location'],axis=1,inplace=True)
```

```python
#Az üres értékkel rendelkező sorok elhagyása

taxi_trips.dropna(inplace=True)
```

```python
taxi_trips.info()
```

## Oszlopok átnevezése

```python
# pickup és dropoff community area oszlopok átnevezése

taxi_trips.rename(columns={'pickup_community_area':'pickup_community_area_id',
                            'dropoff_community_area' : 'dropoff_community_area_id'}, inplace=True )
```

## Transformation: create helper columns for weather and date dimension

```python
# Taxi trips startdate adattípus objectről stringre

taxi_trips['trip_start_timestamp'] = pd.to_datetime(taxi_trips['trip_start_timestamp'])
```

```python
taxi_trips.info()
```

```python
# dátumidő segédoszlop időjáráshoz --> idő lekerekítése a legközelebbi egészórára

taxi_trips['datetime_for_weather'] = taxi_trips["trip_start_timestamp"].dt.floor('h')
```

```python
taxi_trips['datetime_for_weather']
```

```python
# Taxi trips startdate adattípus objectről stringre

taxi_trips['trip_start_date'] = pd.to_datetime(taxi_trips['trip_start_timestamp']).dt.date
```

```python
taxi_trips['trip_start_date'].sample(5)
```

## Taxi és időjárási adatok összekötése

```python
# időjárási adatok beolvasása

current_datetime = datetime.now() - relativedelta(months=2)

formatted_datetime = current_datetime.strftime("%Y-%m-%d")

url = "https://archive-api.open-meteo.com/v1/era5"


# Az URL-beli értékek cseréje dictionary-vel:

params = {
    "latitude": 41.85,
    "longitude": -87.65,
    "start_date": formatted_datetime,
    "end_date": formatted_datetime,
    "hourly" : "temperature_2m,wind_speed_10m,rain,precipitation"
}

response = requests.get(url , params=params)
data = response.json()

weather_data = {
    "datetime" : data["hourly"]["time"],
    "temperature" : data["hourly"]["temperature_2m"],
    "wind_speed" : data["hourly"]["wind_speed_10m"],
    "rain" : data["hourly"]["rain"],
    "precipitation" : data["hourly"]["precipitation"]
}

#Mivel objectnek fordítja alapból a dátumot, ez át kell alakítani

weather_df = pd.DataFrame(weather_data)

weather_df["datetime"] = pd.to_datetime(weather_df["datetime"])

weather_df.head()
```

```python
taxi_trips_with_weather = taxi_trips.merge(weather_df,left_on="datetime_for_weather", right_on="datetime")
```

```python
taxi_trips_with_weather.head()
```

#### Memória használat változása a korábbi és a mostani átalakított df-el

## Create update logic for Dimension tables

```python
dim_payment_type = taxi_trips['payment_type'].drop_duplicates().reset_index(drop=True)

#Majd létrehozzuk a df-et hozzá

dim_payment_type = pd.DataFrame(

    {
        "payment_type_id" : range(1,len(dim_payment_type)+1),
        "payment_type": dim_payment_type
    }
)


# dummy df generálása (mintha a holnapi adatok lennének) --> a kód felkészítése a folyamatos update-re, új adatokra

dummy_payment_type_data = [

{'payment_type': 'Credit Card'},
{'payment_type': 'X'},
{'payment_type': 'Y'},
{'payment_type': 'X'},

]

dummy_payment_type_data_df = pd.DataFrame(dummy_payment_type_data)

#duplikációk eltávolítása

todays_payment_types = pd.DataFrame(dummy_payment_type_data_df['payment_type'].unique(), columns=['payment_type'])

#majd kiadatjuk az összes olyan értéket a dummy táblából ami a korábban definiált dim táblában nincs benne

new_payment_types = todays_payment_types[~todays_payment_types['payment_type'].isin(dim_payment_type['payment_type'])]

#majd azért, hogy ne a dimenzió táblában lévő számnál lévő értéket írja felül, hanem adjon hozzá egy új ID-t, ezért szükséges az alábbi

if not new_payment_types.empty:
    max_id = dim_payment_type['payment_type_id'].max()
    new_payment_types['payment_type_id'] = range(max_id+1,max_id+1+len(new_payment_types))
    dim_payment_type = pd.concat([dim_payment_type, new_payment_types], ignore_index = True )

dim_payment_type

```

```python
dim_company = taxi_trips['company'].drop_duplicates().reset_index(drop=True)

#Majd létrehozzuk a df-et hozzá

dim_company = pd.DataFrame(

    {
        "company_id" : range(1,len(dim_company)+1),
        "company": dim_company
    }
)


# dummy df generálása (mintha a holnapi adatok lennének) --> a kód felkészítése a folyamatos update-re, új adatokra

dummy_company_data = [

{'company': 'Metro Jet Taxi A'},
{'company': 'X'},
{'company': 'Y'},
{'company': 'X'},

]

dummy_company_data_df = pd.DataFrame(dummy_company_data)

#duplikációk eltávolítása

todays_companies = pd.DataFrame(dummy_company_data_df['company'].unique(), columns=['company'])

#majd kiadatjuk az összes olyan értéket a dummy táblából ami a korábban definiált dim táblában nincs benne

new_companies = todays_companies[~todays_companies['company'].isin(dim_company['company'])]

#majd azért, hogy ne a dimenzió táblában lévő számnál lévő értéket írja felül, hanem adjon hozzá egy új ID-t, ezért szükséges az alábbi

if not new_companies.empty:
    max_id = dim_company['company_id'].max()
    new_companies['company_id'] = range(max_id+1,max_id+1+len(new_companies))
    dim_company = pd.concat([dim_company, new_companies], ignore_index = True )

dim_company
```
